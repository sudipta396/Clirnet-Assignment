````python
import requests
from bs4 import BeautifulSoup

def scrape_page(url):
    try:
        headers = {
            "User-Agent": "Mozilla/5.0"
        }

        response = requests.get(
            url,
            headers=headers,
            timeout=20
        )

        response.raise_for_status()

        soup = BeautifulSoup(response.text, "html.parser")

        # Remove unnecessary elements
        for element in soup([
            "script",
            "style",
            "nav",
            "footer",
            "header",
            "aside"
        ]):
            element.decompose()

        text = soup.get_text(separator=" ", strip=True)

        return text

    except Exception as e:
        print(f"Error scraping {url}: {e}")
        return ""
````

````python
import time

def clean_text(text, max_chars=30000):
    # Remove excessive whitespace
    text = " ".join(text.split())

    # Limit text size for LLM processing
    return text[:max_chars]


documents = []

for i, url in enumerate(urls, start=1):
    print(f"Scraping {i}/10...")

    text = scrape_page(url)
    text = clean_text(text)

    documents.append({
        "link": url,
        "text": text
    })

    print(f"  Characters kept: {len(text)}")

    # Small delay between requests
    time.sleep(1)

print("\nFinished scraping all URLs.")
print("Documents:", len(documents))
````

````python
from typing import List
from pydantic import BaseModel, Field
import dspy


class EntityWithAttr(BaseModel):
    entity: str = Field(
        description="The exact named entity or concept from the text"
    )

    attr_type: str = Field(
        description="Semantic type such as Drug, Disease, Crop, Process, Organization, Measurement, etc."
    )


class ExtractEntities(dspy.Signature):
    """Extract important named entities and concepts from the given paragraph."""

    paragraph: str = dspy.InputField()

    entities: List[EntityWithAttr] = dspy.OutputField()
````

````python
import re
from collections import Counter


def local_entity_extraction(text):
    """
    Temporary fallback extractor.
    Used only to build/test the downstream pipeline
    while the LongCat API quota is unavailable.
    """

    # Common domain-specific terms
    patterns = [
        r'\b(?:sustainable agriculture|organic farming|precision agriculture|'
        r'climate change|soil health|water management|crop rotation|'
        r'crop diversification|agroforestry|biodiversity|irrigation)\b',

        r'\b(?:COVID-19|cancer|diabetes|tramadol|shingles|'
        r'vaccine|disease|pain|health)\b',

        r'\b(?:NASA|WHO|FAO|FDA|ESA)\b'
    ]

    found = []

    for pattern in patterns:
        found.extend(re.findall(pattern, text, flags=re.IGNORECASE))

    # Normalize duplicates while preserving exact first occurrence
    unique = []
    seen = set()

    for entity in found:
        entity = entity.strip()

        key = entity.lower()

        if key not in seen:
            seen.add(key)
            unique.append(entity)

    return unique


# Test
test_entities = local_entity_extraction(documents[0]["text"])

print("Entities found:", len(test_entities))
print(test_entities[:20])
````

````python
def classify_entity(entity):
    """
    Temporary semantic classifier.
    This will eventually be replaced/verified using the LLM.
    """

    e = entity.lower()

    if any(x in e for x in [
        "agriculture", "farming", "agroforestry",
        "crop rotation", "irrigation"
    ]):
        return "Process"

    if any(x in e for x in [
        "soil health", "biodiversity", "water management"
    ]):
        return "Environmental Concept"

    if any(x in e for x in [
        "covid", "cancer", "diabetes", "disease",
        "pain", "shingles"
    ]):
        return "Disease/Health"

    if any(x in e for x in [
        "tramadol", "vaccine"
    ]):
        return "Drug/Medical"

    if entity.upper() in ["NASA", "WHO", "FAO", "FDA", "ESA"]:
        return "Organization"

    return "Concept"
````

````python
from difflib import SequenceMatcher


def similarity(a, b):
    return SequenceMatcher(
        None,
        a.lower(),
        b.lower()
    ).ratio()


def deduplicate_entities(entities, threshold=0.85):
    """
    Remove highly similar duplicate entities.
    Keeps the first occurrence.
    """

    unique = []

    for entity in entities:
        entity = entity.strip()

        if not entity:
            continue

        is_duplicate = False

        for existing in unique:
            if similarity(entity, existing) >= threshold:
                is_duplicate = True
                break

        if not is_duplicate:
            unique.append(entity)

    return unique
````

````python
from itertools import combinations


def generate_relationships(entities):
    """
    Temporary relationship generator.
    Creates relationships between entities appearing
    in the same document.
    """

    relationships = []

    for source, target in combinations(entities, 2):
        relationships.append({
            "source": source,
            "target": target,
            "relation": "related to"
        })

    return relationships
````

````python
def clean_mermaid_text(text):
    """
    Make text safer for Mermaid syntax.
    """

    text = str(text).replace('"', "'")
    text = " ".join(text.split())

    return text[:40]


def triples_to_mermaid(triples, entity_list):
    """
    Convert relationships into a Mermaid graph.
    Only entities in entity_list can become nodes.
    """

    valid_entities = set(entity_list)

    lines = ["graph LR"]

    # Add nodes
    for i, entity in enumerate(entity_list):
        node_id = f"N{i}"
        safe_entity = entity.replace('"', "'")

        lines.append(
            f'    {node_id}["{safe_entity}"]'
        )

    # Map entity → node ID
    entity_to_id = {
        entity: f"N{i}"
        for i, entity in enumerate(entity_list)
    }

    # Add edges
    for triple in triples:

        source = triple["source"]
        target = triple["target"]

        if source not in valid_entities:
            continue

        if target not in valid_entities:
            continue

        relation = clean_mermaid_text(
            triple["relation"]
        )

        lines.append(
            f'    {entity_to_id[source]} -->|"{relation}"| '
            f'{entity_to_id[target]}'
        )

    return "\n".join(lines)
````

````python
def extract_entities_dspy(text, chunk_size=8000):
    """
    Extract structured entities from text using DSPy.

    Text is processed in chunks to avoid excessively large
    LLM requests.
    """

    all_entities = []

    for start in range(0, len(text), chunk_size):
        chunk = text[start:start + chunk_size]

        try:
            result = entity_extractor(
                paragraph=chunk
            )

            if result.entities:
                all_entities.extend(result.entities)

        except Exception as e:
            print(f"Extraction error: {e}")

    return all_entities
````

````python
def normalize_entities(entities):
    """
    Convert Pydantic entity objects into clean dictionaries
    and remove exact duplicates.
    """

    unique = {}
    for item in entities:

        entity = item.entity.strip()
        attr_type = item.attr_type.strip()

        if not entity:
            continue

        key = entity.lower()

        if key not in unique:
            unique[key] = {
                "entity": entity,
                "attr_type": attr_type
            }

    return list(unique.values())
````

````python
def process_document_dspy(text):
    """
    Complete entity extraction pipeline:

    text
      ↓
    DSPy extraction
      ↓
    normalization
      ↓
    deduplication
    """

    entities = extract_entities_dspy(text)

    entities = normalize_entities(entities)

    return entities
````

````python
from pydantic import BaseModel, Field
from typing import List


class Relationship(BaseModel):
    source: str = Field(
        description="Source entity"
    )

    relation: str = Field(
        description="Short semantic relationship between the entities"
    )

    target: str = Field(
        description="Target entity"
    )


class ExtractRelationships(dspy.Signature):
    """Extract meaningful relationships between the provided entities."""

    paragraph: str = dspy.InputField()

    entity_list: str = dspy.InputField(
        description="List of allowed entities. Use only these entities."
    )

    relationships: List[Relationship] = dspy.OutputField()
````

````python
relationship_extractor = dspy.Predict(
    ExtractRelationships
)

print("Relationship extractor created successfully!")
````

````python
def extract_relationships_dspy(text, entities):
    """
    Extract relationships using only the supplied
    deduplicated entities.
    """

    entity_names = [
        item["entity"] if isinstance(item, dict)
        else item.entity
        for item in entities
    ]

    entity_text = ", ".join(entity_names)

    try:
        result = relationship_extractor(
            paragraph=text[:8000],
            entity_list=entity_text
        )

        relationships = []

        for r in result.relationships:

            if (
                r.source in entity_names
                and r.target in entity_names
                and r.source != r.target
            ):
                relationships.append({
                    "source": r.source,
                    "relation": r.relation[:40].strip(),
                    "target": r.target
                })

        return relationships

    except Exception as e:
        print("Relationship extraction error:", e)
        return []
````

````python
def process_document(link, text):
    """
    Complete pipeline for one webpage.

    1. Extract entities with DSPy
    2. Normalize entities
    3. Deduplicate entities
    4. Extract relationships
    5. Generate Mermaid graph
    """

    print(f"\nProcessing: {link}")

    # 1. Entity extraction
    entities = extract_entities_dspy(text)

    # 2. Normalize
    entities = normalize_entities(entities)

    print(f"Entities extracted: {len(entities)}")

    # 3. Deduplicate
    entity_names = [e["entity"] for e in entities]
    unique_names = deduplicate_entities(entity_names)

    # Keep corresponding types
    unique_entities = []

    for name in unique_names:
        for entity in entities:
            if entity["entity"].lower() == name.lower():
                unique_entities.append(entity)
                break

    print(f"After deduplication: {len(unique_entities)}")

    # 4. Relationship extraction
    relationships = extract_relationships_dspy(
        text,
        unique_entities
    )

    print(f"Relationships: {len(relationships)}")

    # 5. Mermaid generation
    mermaid = triples_to_mermaid(
        relationships,
        [e["entity"] for e in unique_entities]
    )

    return {
        "link": link,
        "entities": unique_entities,
        "relationships": relationships,
        "mermaid": mermaid
    }
````

````python
import os
import pandas as pd

os.makedirs("final_output/mermaid", exist_ok=True)


def save_result(result, index):
    """
    Save CSV records and Mermaid file for one document.
    """

    # Save Mermaid
    mermaid_path = (
        f"final_output/mermaid/mermaid_{index}.md"
    )

    with open(
        mermaid_path,
        "w",
        encoding="utf-8"
    ) as f:

        f.write("```mermaid\n")
        f.write(result["mermaid"])
        f.write("\n```\n")

    # Create CSV records
    rows = []

    for entity in result["entities"]:
        rows.append({
            "link": result["link"],
            "tag": entity["entity"],
            "tag_type": entity["attr_type"]
        })

    return rows
````

````python
import requests
from bs4 import BeautifulSoup

def scrape_sciencedirect_fallback(url):
    headers = {
        "User-Agent": "Mozilla/5.0 (X11; Linux x86_64) AppleWebKit/537.36 Chrome/139.0 Safari/537.36",
        "Accept": "text/html,application/xhtml+xml,application/xml;q=0.9,*/*;q=0.8",
        "Accept-Language": "en-US,en;q=0.9",
    }

    try:
        response = requests.get(
            url,
            headers=headers,
            timeout=30
        )

        print("Status:", response.status_code)

        soup = BeautifulSoup(response.text, "html.parser")

        # Look for article metadata
        title = soup.find("meta", attrs={"name": "citation_title"})
        abstract = soup.find("meta", attrs={"name": "description"})

        title_text = title.get("content", "") if title else ""
        abstract_text = abstract.get("content", "") if abstract else ""

        text = f"{title_text} {abstract_text}".strip()

        return text

    except Exception as e:
        print("Error:", e)
        return ""
````

````python
def run_pipeline(documents):
    all_rows = []

    os.makedirs("final_output/mermaid", exist_ok=True)

    for i, doc in enumerate(documents, start=1):

        # Skip documents with no text
        if not doc["text"].strip():
            print(f"Skipping URL {i}: no text available")
            continue

        result = process_document(
            doc["link"],
            doc["text"]
        )

        rows = save_result(result, i)
        all_rows.extend(rows)

    # Create final CSV
    final_df = pd.DataFrame(
        all_rows,
        columns=["link", "tag", "tag_type"]
    )

    final_df = final_df.drop_duplicates(
        subset=["link", "tag"]
    ).reset_index(drop=True)

    final_df.to_csv(
        "final_output/tags.csv",
        index=False
    )

    return final_df
````
